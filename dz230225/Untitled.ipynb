```python
#ЗАДАНИЕ ЧЕТЫРЕ
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint

def f(xyz, t, s=10, r=28, b=8/3):
    # функция возвращает значения производных в заданной точке
    x, y, z = xyz
    x_dot = s*(y - x)
    y_dot = r*x - y - x*z
    z_dot = x*y - b*z
    return np.array([x_dot, y_dot, z_dot])

t = np.linspace(0,20,2000)
v0 = (1., -1., 10)  # инициализируем начальными значениями
v02 = (1.001, -1., 10)

xyzs =odeint(f, v0, t, full_output=False)
xyzs2 =odeint(f, v02, t, full_output=False)


# Plot
ax = plt.figure().add_subplot(projection='3d')

# .T транспонирование  (колонки станут строками, а строки - колонками)
ax.plot(*xyzs.T, lw=0.5)  # lw ширина линии

ax.plot(*xyzs2.T, lw=0.5)  # lw ширина линии

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
ax.set_title("Аттрактор Лоренца")

plt.show()
```

```python
#ЗАДАНИЕ >= пять
import math


import matplotlib.pyplot as plt


import numpy as np


import random

from scipy.integrate import odeint

maxd = 20

a = []

for i in range(2000):
    x1 = random.randint(0, maxd * 100) /100
    x2 = random.randint(0, maxd * 100) /100
    a.append([x1, x2, abs(x1 - x2)])
gs = {}
def f(xyz, t, s=10, r=28, b=8/3):
    # функция возвращает значения производных в заданной точке
    x, y, z = xyz
    x_dot = s*(y - x)
    y_dot = r*x - y - x*z
    z_dot = x*y - b*z
    return np.array([x_dot, y_dot, z_dot])

t = np.linspace(0,20,2000)
for i in a:
    v0 = (i[0], -1., 10)  # инициализируем начальными значениями
    v02 = (i[1], -1., 10)

    xyzs = odeint(f, v0, t, full_output=False)
    xyzs2 = odeint(f, v02, t, full_output=False)
    mind = 10000
    maxd = 0
    for j in range(len(xyzs)):
        r = round(i[2])
        d = math.sqrt((xyzs[j, 0] - xyzs2[j, 0]) ** 2 + (xyzs[j, 1] - xyzs2[j, 1]) ** 2 + (xyzs[j, 2] - xyzs2[j, 2]) ** 2)
        maxd = max(maxd, d)
        mind = min(mind, d)
    if r not in gs.keys():
        gs.update({r: {"is": [i],"mins":[mind], "maxs" : [maxd]}})
    else:
        gs[r]["is"].append(i)
        gs[r]["mins"].append(mind)
        gs[r]["maxs"].append(maxd)

ax = plt.figure().add_subplot()
p1 = []
p2 = []
p3 = []

p12 = []
p22 = []
p32 = []

for i in gs.keys():
    p1.append([i, max(gs[i]["maxs"])])
    p2.append([i, sum(gs[i]["maxs"])/len(gs[i]["maxs"])])
    p3.append([i, min(gs[i]["maxs"])])

    p12.append([i, max(gs[i]["mins"])])
    p22.append([i, sum(gs[i]["mins"]) / len(gs[i]["mins"])])
    p32.append([i, min(gs[i]["mins"])])

ax.plot(p1, label="max max")
ax.plot(p2, label="sred max")
ax.plot(p3, label="min max")

ax.plot(p12, label="max min")
ax.plot(p22, label="sred min")
ax.plot(p32, label="min min")

plt.show()
```
