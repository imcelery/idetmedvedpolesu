```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import numpy as np
import math


class Animation:
    def __init__(self, width, height):
        self.width = width
        self.height = height

        self.planets = []

    
    def add_planet(self,  color, orbit, speed):
        self.planets.append({ "color": color,  "plot":None, "angel" : 0, "speed" : speed, "orbit":orbit})
        



    def show(self, plot):
        gs = []

            
        cx = width // 2
        cy = height // 2
        
        for i in range(len(self.planets)):
            xdata = []
            ydata = []
            self.planets[i]["angel"] += self.planets[i]["speed"]
            t = self.planets[i]
            posx = math.cos(math.radians(t["angel"])) * t["orbit"] + cx
            posy = math.sin(math.radians(t["angel"])) * t["orbit"] + cy
            for j in range(360):
                xdata.append(math.cos(math.radians(j)) * 10 + posx)
                ydata.append(math.sin(math.radians(j)) * 10 + posy)

            if t["plot"] is None:
                t["plot"] = plot.plot(xdata, ydata, c=t["color"])
            else:
                t["plot"][0].set_data(xdata, ydata)
            gs.append(t["plot"][0])


        return tuple(gs)


width = 3000
height = 3000

a = Animation(width, height)


#planets
a.add_planet('yellow', 0, 0)
a.add_planet('red', 20, 47)
a.add_planet('orange', 35, 35)
a.add_planet('green', 50, 30)
a.add_planet('red', 75, 24)

a.add_planet('red', 260, 13)
a.add_planet('orange', 475, 10)

a.add_planet('blue', 960, 7)
a.add_planet('blue', 1500, 6)
a.add_planet('gray', 1950, 5)



fig = plt.figure(figsize=(20,20))
ax = plt.axes(xlim=(0, width), ylim=(0, height))

ax.set_xlim([0, width])
ax.set_ylim([0, height])


def init():
    a.set_k("c", "y", 0)
    return a.show(ax)


def animate(frame):
    return a.show(ax)

anim = animation.FuncAnimation(fig, animate,  frames=200, interval=1 )
anim.save(filename = "planet.gif")
```
